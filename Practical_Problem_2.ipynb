```python
import pandas as pd

df = pd.read_csv('/content/enjoysport - enjoysport.csv')
display(df.head())
```

```python
import numpy as np

def candidate_elimination(df):

    X = df.iloc[:, :-1].values
    y = df.iloc[:, -1].values
    n = X.shape[1]

    # Initial hypotheses
    S = ['0'] * n
    G = [['?'] * n]

    for i in range(len(X)):

        print(f"\nExample {i+1}: {list(X[i])} -> {y[i]}")

        if y[i] == 'yes':

            # Update Specific hypothesis
            for j in range(n):
                if S[j] == '0':
                    S[j] = X[i][j]
                elif S[j] != X[i][j]:
                    S[j] = '?'

            # Remove G hypotheses that don't cover S
            G = [g for g in G if all(
                g[j] == '?' or g[j] == S[j]
                for j in range(n)
            )]

        else:

            # Specialize G for negative example
            new_G = []

            for g in G:
                for j in range(n):
                    if g[j] == '?' and S[j] != '?':
                        new_g = g.copy()
                        new_g[j] = S[j]

                        if new_g not in new_G:
                            new_G.append(new_g)

            G = new_G

        print("SPECIFIC HYPOTHESIS =", S)
        print("GENERAL HYPOTHESIS =", G)

    return S, G


# Run Candidate Elimination
S_final, G_final = candidate_elimination(df)

print("\nFinal Specific Hypothesis:")
print(S_final)

print("\nFinal General Hypotheses:")
print(G_final)
```
