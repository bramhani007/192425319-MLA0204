```python
import pandas as pd

df = pd.read_csv('/content/enjoysport - enjoysport.csv')
display(df.head())
```

```python
import numpy as np

def candidate_elimination(data):
    features = data.iloc[:, :-1].values
    target = data.iloc[:, -1].values
    num_attributes = features.shape[1]


    specific_hypothesis = ['0'] * num_attributes

    general_hypotheses = [['?'] * num_attributes]

    print("\nInitial Specific Hypothesis (S0):", specific_hypothesis)
    print("Initial General Hypothesis (G0):", [np.array(h, dtype=object) for h in general_hypotheses])


    def covers(hypothesis_to_check, example_instance):
        for j in range(num_attributes):
            if hypothesis_to_check[j] != '?' and hypothesis_to_check[j] != example_instance[j]:
                return False
        return True

    for i, (instance, label) in enumerate(zip(features, target)):
        print(f"\n\nTraining Example {i + 1}: {list(instance)}, Target: {label}")

        if label == 'yes':

            general_hypotheses = [g for g in general_hypotheses if covers(g, instance)]


            if specific_hypothesis == ['0'] * num_attributes:
                specific_hypothesis = list(instance)
            else:
                for j in range(num_attributes):

                    if specific_hypothesis[j] != instance[j]:
                        specific_hypothesis[j] = '?'


            general_hypotheses = [g for g in general_hypotheses if covers(g, specific_hypothesis)]

        else:

            new_general_hypotheses = []
            for g in general_hypotheses:
                if covers(g, instance):
                    for j in range(num_attributes):

                        if g[j] == '?' and specific_hypothesis[j] != '?' and specific_hypothesis[j] != instance[j]:
                            new_g = list(g)
                            new_g[j] = specific_hypothesis[j]


                            if covers(new_g, specific_hypothesis) and new_g not in new_general_hypotheses:
                                new_general_hypotheses.append(new_g)
                else:
                    new_general_hypotheses.append(g)


            unique_general_hypotheses = []
            seen = set()
            for h_list in new_general_hypotheses:
                h_tuple = tuple(h_list)
                if h_tuple not in seen and covers(h_list, specific_hypothesis):
                    unique_general_hypotheses.append(h_list)
                    seen.add(h_tuple)
            general_hypotheses = unique_general_hypotheses


        print(f"Specific Hypothesis (S{i+1}):", specific_hypothesis)
        print(f"General Hypotheses (G{i+1}):", [np.array(h, dtype=object) for h in general_hypotheses])

    return specific_hypothesis, general_hypotheses


s_final, g_final = candidate_elimination(df.copy())

print("\n\nFinal Specific Hypothesis (S):", s_final)
print("Final General Hypotheses (G):", [np.array(h, dtype=object) for h in g_final])
```
