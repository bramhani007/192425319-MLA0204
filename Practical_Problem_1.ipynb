```python
import pandas as pd

df = pd.read_csv('/content/enjoysport - enjoysport.csv')
display(df.head())
```

```python
def find_s(df):

    # Separate attributes and target
    X = df.iloc[:, :-1].values
    y = df.iloc[:, -1].values

    # Find first positive example
    for i in range(len(y)):
        if y[i].lower() == "yes":
            hypothesis = list(X[i])
            break

    print("Initial Hypothesis:", hypothesis)

    # Process positive examples
    for i in range(len(y)):

        if y[i].lower() == "yes":

            for j in range(len(hypothesis)):

                if hypothesis[j] != X[i][j]:
                    hypothesis[j] = "?"

            print("After Example", i + 1, ":", hypothesis)

    return hypothesis


# Apply FIND-S
final_hypothesis = find_s(df)

print("\nFinal Most Specific Hypothesis:")
print(final_hypothesis)
```
